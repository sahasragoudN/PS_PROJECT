```python
# ============================================
# STEP 4: CORRECT DAVIS Y TRANSFORMATION
# ============================================

import math
import numpy as np
import pickle
import json
import os

DAVIS_DIR = "/content/My_DeepDTA/data/raw/davis"

# Load original Davis Y
with open(os.path.join(DAVIS_DIR, "Y"), "rb") as f:
    Y_davis_raw = pickle.load(f, encoding="latin1")

Y_davis_raw = np.array(Y_davis_raw, dtype=np.float32)

print("Original Y shape:", Y_davis_raw.shape)

# Davis pKd transformation
# pKd = -log10(Kd / 1e9)

Y = np.full_like(Y_davis_raw, np.nan, dtype=np.float32)

valid = ~np.isnan(Y_davis_raw)

Y[valid] = -np.log10(
    Y_davis_raw[valid] / 1e9
)

print("Transformed Y shape:", Y.shape)

print("\nMinimum pKd:", np.nanmin(Y))
print("Maximum pKd:", np.nanmax(Y))
print("Mean pKd:", np.nanmean(Y))

print("\nFirst 5 valid values:")
print(Y[valid][:5])
```

```python
# ============================================
# STEP 5: CORRECT DAVIS TRAIN / TEST FOLDS
# ============================================

import json
import numpy as np
import os

DAVIS_DIR = "/content/My_DeepDTA/data/raw/davis"

# Find all valid (non-NaN) entries in Y
label_row_inds, label_col_inds = np.where(
    np.isnan(Y) == False
)

print("Total valid Davis interactions:", len(label_row_inds))


# Load official test fold
with open(
    os.path.join(DAVIS_DIR, "folds", "test_fold_setting1.txt"),
    "r"
) as f:
    test_fold = json.load(f)


# Load official training folds
with open(
    os.path.join(DAVIS_DIR, "folds", "train_fold_setting1.txt"),
    "r"
) as f:
    train_folds = json.load(f)


print("Number of training folds:", len(train_folds))
print("Test samples:", len(test_fold))


# Use the FIRST official training fold
train_fold = np.array(train_folds[0])
test_fold = np.array(test_fold)


# Convert fold positions into drug/protein indices
train_drug_indices = label_row_inds[train_fold]
train_protein_indices = label_col_inds[train_fold]

test_drug_indices = label_row_inds[test_fold]
test_protein_indices = label_col_inds[test_fold]


# Get affinity values
Y_train = Y[
    train_drug_indices,
    train_protein_indices
]

Y_test = Y[
    test_drug_indices,
    test_protein_indices
]


# Display results
print("\n========== DAVIS DATA ==========")

print("Training samples:", len(Y_train))
print("Test samples:", len(Y_test))

print("\nTraining target range:")
print(np.min(Y_train), "to", np.max(Y_train))

print("\nTest target range:")
print(np.min(Y_test), "to", np.max(Y_test))

print("\nFirst 5 training targets:")
print(Y_train[:5])
```

```python
# ============================================
# STEP 6: ENCODE SMILES AND PROTEIN SEQUENCES
# ============================================

# Maximum lengths used for DeepDTA
MAX_SMILES_LENGTH = 100
MAX_PROTEIN_LENGTH = 1000

# --------------------------------------------
# Character dictionaries
# --------------------------------------------

SMILES_CHARS = "#%()+-./0123456789=ABCDEFGHIJKLMNOPQRSTUVWXYZ[]_abcdefghijklmnopqrstuvwxyz"

PROTEIN_CHARS = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

smiles_to_int = {
    char: index + 1
    for index, char in enumerate(SMILES_CHARS)
}

protein_to_int = {
    char: index + 1
    for index, char in enumerate(PROTEIN_CHARS)
}

print("Number of SMILES characters:", len(smiles_to_int))
print("Number of protein characters:", len(protein_to_int))


# --------------------------------------------
# SMILES encoding function
# --------------------------------------------

def encode_smiles(smiles):
    encoded = []

    for char in smiles:
        if char in smiles_to_int:
            encoded.append(smiles_to_int[char])
        else:
            encoded.append(0)

    # Cut if longer than maximum length
    encoded = encoded[:MAX_SMILES_LENGTH]

    # Padding
    encoded += [0] * (MAX_SMILES_LENGTH - len(encoded))

    return np.array(encoded, dtype=np.int32)


# --------------------------------------------
# Protein encoding function
# --------------------------------------------

def encode_protein(sequence):
    encoded = []

    for char in sequence:
        if char in protein_to_int:
            encoded.append(protein_to_int[char])
        else:
            encoded.append(0)

    # Cut if longer than maximum length
    encoded = encoded[:MAX_PROTEIN_LENGTH]

    # Padding
    encoded += [0] * (MAX_PROTEIN_LENGTH - len(encoded))

    return np.array(encoded, dtype=np.int32)


# --------------------------------------------
# Test the encoding
# --------------------------------------------

example_drug_id = list(drugs.keys())[0]
example_protein_id = list(proteins.keys())[0]

example_smiles = drugs[example_drug_id]
example_sequence = proteins[example_protein_id]

encoded_drug = encode_smiles(example_smiles)
encoded_protein = encode_protein(example_sequence)

print("\nExample SMILES:")
print(example_smiles)

print("\nEncoded SMILES:")
print(encoded_drug)

print("\nEncoded SMILES shape:")
print(encoded_drug.shape)

print("\nExample protein sequence:")
print(example_sequence[:50], "...")

print("\nEncoded protein shape:")
print(encoded_protein.shape)
```

```python
# ============================================
# STEP 7: CREATE CORRECT DAVIS INPUT ARRAYS
# ============================================

# --------------------------------------------
# Encode all drugs
# --------------------------------------------

print("Encoding Davis drugs...")

drug_ids = list(drugs.keys())

encoded_drugs = {}

for i, drug_id in enumerate(drug_ids):
    encoded_drugs[i] = encode_smiles(drugs[drug_id])

print("Drugs encoded:", len(encoded_drugs))


# --------------------------------------------
# Encode all proteins
# --------------------------------------------

print("Encoding Davis proteins...")

protein_ids = list(proteins.keys())

encoded_proteins = {}

for i, protein_id in enumerate(protein_ids):
    encoded_proteins[i] = encode_protein(proteins[protein_id])

print("Proteins encoded:", len(encoded_proteins))


# --------------------------------------------
# Create training input arrays
# --------------------------------------------

X_drug_train = np.array([
    encoded_drugs[i]
    for i in train_drug_indices
], dtype=np.int32)

X_protein_train = np.array([
    encoded_proteins[i]
    for i in train_protein_indices
], dtype=np.int32)


# --------------------------------------------
# Create test input arrays
# --------------------------------------------

X_drug_test = np.array([
    encoded_drugs[i]
    for i in test_drug_indices
], dtype=np.int32)

X_protein_test = np.array([
    encoded_proteins[i]
    for i in test_protein_indices
], dtype=np.int32)


# --------------------------------------------
# Make sure targets are float32
# --------------------------------------------

Y_train = np.array(Y_train, dtype=np.float32)
Y_test = np.array(Y_test, dtype=np.float32)


# --------------------------------------------
# Check everything
# --------------------------------------------

print("\n========== DAVIS TRAINING DATA ==========")

print("X_drug_train:    ", X_drug_train.shape)
print("X_protein_train: ", X_protein_train.shape)
print("Y_train:         ", Y_train.shape)

print("\n========== DAVIS TEST DATA ==========")

print("X_drug_test:     ", X_drug_test.shape)
print("X_protein_test:  ", X_protein_test.shape)
print("Y_test:          ", Y_test.shape)

print("\n========== TARGET CHECK ==========")

print("Y_train minimum:", np.min(Y_train))
print("Y_train maximum:", np.max(Y_train))
print("Y_train mean:", np.mean(Y_train))
```

```python
# ============================================
# STEP 8: BUILD DEEPDTA MODEL FOR DAVIS
# ============================================

import tensorflow as tf

from tensorflow.keras.layers import (
    Input,
    Embedding,
    Conv1D,
    GlobalMaxPooling1D,
    Concatenate,
    Dense,
    Dropout
)

from tensorflow.keras.models import Model


# ============================================
# DRUG / SMILES BRANCH
# ============================================

drug_input = Input(
    shape=(MAX_SMILES_LENGTH,),
    name="drug_input"
)

drug_embedding = Embedding(
    input_dim=len(smiles_to_int) + 1,
    output_dim=128,
    name="drug_embedding"
)(drug_input)

drug_conv1 = Conv1D(
    filters=32,
    kernel_size=4,
    activation="relu"
)(drug_embedding)

drug_conv2 = Conv1D(
    filters=64,
    kernel_size=4,
    activation="relu"
)(drug_conv1)

drug_conv3 = Conv1D(
    filters=96,
    kernel_size=4,
    activation="relu"
)(drug_conv2)

drug_pool = GlobalMaxPooling1D()(drug_conv3)


# ============================================
# PROTEIN BRANCH
# ============================================

protein_input = Input(
    shape=(MAX_PROTEIN_LENGTH,),
    name="protein_input"
)

protein_embedding = Embedding(
    input_dim=len(protein_to_int) + 1,
    output_dim=128,
    name="protein_embedding"
)(protein_input)

protein_conv1 = Conv1D(
    filters=32,
    kernel_size=8,
    activation="relu"
)(protein_embedding)

protein_conv2 = Conv1D(
    filters=64,
    kernel_size=8,
    activation="relu"
)(protein_conv1)

protein_conv3 = Conv1D(
    filters=96,
    kernel_size=8,
    activation="relu"
)(protein_conv2)

protein_pool = GlobalMaxPooling1D()(protein_conv3)


# ============================================
# COMBINE DRUG + PROTEIN
# ============================================

combined = Concatenate()([
    drug_pool,
    protein_pool
])


# ============================================
# FULLY CONNECTED LAYERS
# ============================================

dense1 = Dense(
    1024,
    activation="relu"
)(combined)

dropout1 = Dropout(0.1)(dense1)

dense2 = Dense(
    1024,
    activation="relu"
)(dropout1)

dropout2 = Dropout(0.1)(dense2)

dense3 = Dense(
    512,
    activation="relu"
)(dropout2)


# ============================================
# OUTPUT
# ============================================

output = Dense(
    1,
    activation="linear",
    name="affinity_output"
)(dense3)


# ============================================
# CREATE MODEL
# ============================================

model = Model(
    inputs=[drug_input, protein_input],
    outputs=output,
    name="DeepDTA_Davis"
)


# ============================================
# COMPILE
# ============================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)


print("DeepDTA Davis model created successfully!")

model.summary()
```

```python
# ============================================
# STEP 9: TRAIN DAVIS DEEPDTA
# ============================================

from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    [X_drug_train, X_protein_train],
    Y_train,

    validation_split=0.1,

    epochs=20,
    batch_size=256,

    callbacks=[early_stopping],

    verbose=1
)

print("\nDavis training completed!")
```

```python
# ============================================
# STEP 10: EVALUATE THE DEEPDTA MODEL
# ============================================

import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Predict on the Davis test set
Y_pred = model.predict(
    [X_drug_test, X_protein_test],
    verbose=1
).flatten()

# Calculate evaluation metrics
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)

print("\n========== DAVIS TEST RESULTS ==========")
print("MSE  :", mse)
print("RMSE :", rmse)
print("MAE  :", mae)
print("R²   :", r2)

print("\n========== SAMPLE PREDICTIONS ==========")

for i in range(10):
    print(
        f"Actual: {Y_test[i]:.4f}   "
        f"Predicted: {Y_pred[i]:.4f}"
    )
```

```python
# ============================================
# STEP 12: CREATE DEEPDTA PREDICTION BACKEND
# ============================================

import numpy as np
import tensorflow as tf

# Load the saved trained model
MODEL_PATH = "/content/My_DeepDTA/models/deepdta.keras"

trained_model = tf.keras.models.load_model(MODEL_PATH)

print("Trained DeepDTA model loaded successfully!")


# --------------------------------------------
# Prediction function
# --------------------------------------------

def predict_affinity(smiles, protein_sequence):

    # Encode drug SMILES
    drug_encoded = encode_smiles(smiles)

    # Encode protein sequence
    protein_encoded = encode_protein(protein_sequence)

    # Add batch dimension
    drug_input = np.expand_dims(drug_encoded, axis=0)
    protein_input = np.expand_dims(protein_encoded, axis=0)

    # Make prediction
    prediction = trained_model.predict(
        [drug_input, protein_input],
        verbose=0
    )

    # Extract predicted pKd value
    predicted_pkd = float(prediction[0][0])

    return predicted_pkd


print("Prediction function created successfully!")
```

```python
BACKEND_DIR = "/content/My_DeepDTA/backend"
import os

os.makedirs(BACKEND_DIR, exist_ok=True)

app_code = """
from fastapi import FastAPI
from pydantic import BaseModel
import numpy as np
import tensorflow as tf
import os

# --------------------------------------------------
# Create FastAPI application
# --------------------------------------------------

app = FastAPI(
    title="DeepDTA API",
    description="Drug-Target Binding Affinity Prediction API",
    version="1.0"
)


# --------------------------------------------------
# Load trained DeepDTA model
# --------------------------------------------------

MODEL_PATH = "/content/My_DeepDTA/models/deepdta.keras"

model = tf.keras.models.load_model(MODEL_PATH)


# --------------------------------------------------
# Encoding configuration
# --------------------------------------------------

MAX_SMILES_LENGTH = 100
MAX_PROTEIN_LENGTH = 1000

SMILES_CHARS = "#%()+-./0123456789=ABCDEFGHIJKLMNOPQRSTUVWXYZ[]_abcdefghijklmnopqrstuvwxyz"
PROTEIN_CHARS = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

smiles_to_int = {
    char: index + 1
    for index, char in enumerate(SMILES_CHARS)
}

protein_to_int = {
    char: index + 1
    for index, char in enumerate(PROTEIN_CHARS)
}


# --------------------------------------------------
# Encoding functions
# --------------------------------------------------

def encode_smiles(smiles):

    encoded = []

    for char in smiles:

        if char in smiles_to_int:
            encoded.append(smiles_to_int[char])
        else:
            encoded.append(0)

    encoded = encoded[:MAX_SMILES_LENGTH]

    encoded += [0] * (
        MAX_SMILES_LENGTH - len(encoded)
    )

    return np.array(encoded, dtype=np.int32)


def encode_protein(sequence):

    encoded = []

    for char in sequence:

        if char in protein_to_int:
            encoded.append(protein_to_int[char])
        else:
            encoded.append(0)

    encoded = encoded[:MAX_PROTEIN_LENGTH]

    encoded += [0] * (
        MAX_PROTEIN_LENGTH - len(encoded)
    )

    return np.array(encoded, dtype=np.int32)


# --------------------------------------------------
# Request format
# --------------------------------------------------

class PredictionRequest(BaseModel):

    smiles: str
    protein: str


# --------------------------------------------------
# API home
# --------------------------------------------------

@app.get("/")
def home():

    return {
        "message": "DeepDTA API is running",
        "status": "success"
    }


# --------------------------------------------------
# Prediction endpoint
# --------------------------------------------------

@app.post("/predict")
def predict(request: PredictionRequest):

    smiles = request.smiles.strip()
    protein = request.protein.strip().upper()

    # Check empty inputs
    if not smiles:

        return {
            "error": "Drug SMILES cannot be empty."
        }

    if not protein:

        return {
            "error": "Protein sequence cannot be empty."
        }


    # Encode inputs
    drug_encoded = encode_smiles(smiles)
    protein_encoded = encode_protein(protein)


    # Add batch dimension
    drug_input = np.expand_dims(
        drug_encoded,
        axis=0
    )

    protein_input = np.expand_dims(
        protein_encoded,
        axis=0
    )


    # Make prediction
    prediction = model.predict(
        [drug_input, protein_input],
        verbose=0
    )


    predicted_pkd = float(
        prediction[0][0]
    )


    # Return JSON response
    return {
        "prediction": predicted_pkd,
        "unit": "pKd"
    }
"""

with open(
    os.path.join(BACKEND_DIR, "app.py"),
    "w"
) as f:
    f.write(app_code)

print("backend/app.py created successfully!")
```
